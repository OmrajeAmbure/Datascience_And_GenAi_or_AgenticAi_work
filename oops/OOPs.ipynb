### Step 1: Design the Student

- Before writing code, ask yourself: *What information does every student have?*
- Example:
    - Roll Number
    - Name
    - Age
    - Gender
    - Branch
    - Semester
    - Email
    - Phone Number
    - Address
    - Percentage
      
These become the attributes (data) of a Student.

**Attributes in Python are defined very differently than in Java.**
- In Java, you must declare attributes outside methods at the top of the class with strict data types. 
- In Python, attributes are created dynamically—usually inside the constructor method—without any type declarations

```python
# This is the Blueprint or Module or User Defined DataType Or Data Definations
class Student:
    def __init__(self,roll_no,name,age,gender,branch,semester,email,phone_number,address,percentage):
        self.roll_no: int = roll_no
        self.name: str = name
        self.age: int = age
        self.gender: str = gender
        self.branch: str = branch
        self.semester: int = semester
        self.email: str = email
        self.phone_number: str = phone_number
        self.address: str = address
        self.percentage: float = percentage

    def display_Students(self):
        print(f"""
        ==============================================================================
            Roll No : {self.roll_no}
            Name : {self.name}
            Age : {self.age}
            Gender : {self.gender}
            Branch : {self.branch}
            Semester : {self.semester}
            email : {self.email}
            Phone Number : {self.phone_number}
            Address : {self.address}
            Percentage : {self.percentage}
        ===============================================================================
        """)
    
    
# if __name__ == "__main__":
#     # s1 is object we can created 
#     s1 = Student(1,'Omraje Vikas Ambure',21,"Male","AI&DS",7,"omrajeambure22@gmail.com","8799911512","ATP: Dahiwadi,Tal: Tuljapur,Dist: Dharashiv,413624",76.36)
#     s2 = Student(2,'Vjay Samandan Sambare',21,"Male","AI&DS",7,"vijay@gmail.com","8976463823","ATP: Nana Peth,Tal: Haveli,Dist: Pune",90.78)
#     s3 = Student(3,'Rohit Sham Mane',21,"Male","AI&DS",7,"rohit@gmail.com","9167895463","ATP: Kati,Tal: Tuljapur,Dist: Dharashiv,413624",76.36)
#     s4 = Student(4,'Mohit Umesh Ghodake',21,"Male","AI&DS",7,"mohit@gmail.com","9663782812","ATP: Hadepser,Tal: Haveli,Dist: Pune",90.78)
#     s1.display_Students()
#     s2.display_Students()
#     s3.display_Students()
#     s4.display_Students()
```

Now you'll start understanding the relationship between objects and collections of objects, which is a key OOP concept.

```python
if __name__ == "__main__":
    list_records = list()
    while True:
        print("""
            1. Register Student
            2. Display Student Details
            3. Update Student Details
            4. Delete Student
            5. Search Student by ID
            6. Count Total Students
            7. Exit
        """)
        choice = int(input())
        if choice == 1:
            print("How Many Student Want's Register : ")
            n = int(input())
            print(f"Enter The {n} Records : ")
            for i in range(0,n):
                roll_no = input("\nRoll No : ")
                name = input("Name: ")
                age = input("Age: ")
                gender = input("Gender: ")
                branch = input("Branch: ")
                sem = input("Semester: ")
                email = input("Email: ")
                phone_number = input("Phone Number: ")
                address = input("Address: ")
                percentage = input("Percentage: ")
                student_obj = Student(int(roll_no),name,int(age),gender,branch,int(sem),email,phone_number,address,float(percentage))
                list_records.append(student_obj)
        if choice == 2:
            print("Student Records : ")
            for i in list_records:
                print(f"{i.display_Students()}")
        if choice == 5:
            print("Enter The Student ID : ")
            stu_id = int(input())
            for student in list_records:
                if student.roll_no == stu_id:
                    print(student.display_Students())
        if choice == 7:
            break
    print("Thnak You...! Visit Again ")
    print(list_records)
```
